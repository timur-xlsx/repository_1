## **Работа с API**

```python
import requests
import pandas as pd
```

### **Получить информацию обо всех книгах и сохранить в pd.DataFrame**

```python
url_books = "https://www.anapioficeandfire.com/api/books"
response = requests.get(url_books)
books_data = response.json()
books_df = pd.DataFrame(books_data)
books_df[:2]
```

### **Получить информацию обо всех домах Вестероса (Houses) и сохранить в pd.DataFrame**

```python
url_houses = "https://www.anapioficeandfire.com/api/houses"
response = requests.get(url_houses)
houses_data = response.json()
houses_df = pd.DataFrame(houses_data)
houses_df[:2]
```

###  **Получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в pd.DataFrame**

```python
# То же самое, но добавить params
# Можно params дописать к урлу через вопросительный знак
url_houses_titles = "https://www.anapioficeandfire.com/api/houses"
params = {
    'hasTitles': True
}
response = requests.get(url_houses_titles,
                        params=params)
houses_data_titles = response.json()
houses_df_titles = pd.DataFrame(houses_data_titles)
houses_df_titles[:2]
```
